# Task 3 : User Menu (While loop + break/continue)

1. implement a simple while-loop driven menu that keeps askingthe user to choose an action until they press q and quit
 1 - add order amount to a running lsit(using input)
 2- show all orders and total after applying discounts 
 q - Quit

```python
# creating the orderd list
orders = []
# using while driven
while True:
	print("\n1 - Add Order Press 1 To Add Your Order")
	print("2 - Show Orders Press 2 To See Your Orders")
	print("q - Quit Press Q or q To Complete Your Shopping ")
	print("-"*100)
	choice = input("Enter Your Choice : ")

	if choice == "q" or  choice == "Q":
		print("Thank You! For Shopping from us")
		print("-"*100)
# using the break if the user is quit the program it shoud be terminate the loop and end the program
		break
# adding the order to list through the append()
	elif choice == "1":
		try:
			print("Your Choose 1 Please Add the Order Amount")
			print("-"*100)

			amount = int(input("Enter the Order Amount : "))
			orders.append(amount)
			print("Your Order Added ")
		except ValueError:
			print("Error : Please Enter a Valid Number ")
	else:

		print("You Choose the 2 There is Your All The Order With Amounts ")
		total = 0
		for order in orders:
			if order >= 2000:
				discount = 15
			elif 1500 <= order < 2000:
				discount = 10
			elif 1000 <= order < 1500:
				discount = 7
			else:
				discount = 0

			final_amount = order - (order * discount/100)
			print(f"Orders_amount : {order}  -->  Discount : {discount}%  -->  Final_Amount : {final_amount}")

			total += final_amount
		print(f"The Total Amount You have to Pay is : {total}")
```
