# 2. Applying the discount rules and printing the final values

```python
# applying the discount rules

order_amount = int(input("Enter a order amount"))

if order_amount >= 2000:
    discount = 0.15

elif 1500 <= order_amount < 2000:
    discount = 0.10

elif 1000 <= order_amount < 1500:
    discount = 0.07

else:
    discount = 0.0

# calculating the final amount
discount_amount = order_amount * discount
final_amount = order_amount - discount_amount

# printing the final amount 
print(f"Your Original Amount is : {order_amount}")
print(f"Your Discount Amount is : {discount_amount}")
print(f"Your Final Amount to Pay is :{final_amount}")
```
