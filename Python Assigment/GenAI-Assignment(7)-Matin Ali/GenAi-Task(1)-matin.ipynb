# Task 1:- Basic Class And Object Creation

```python
# Creating a class Product
class Product:
	# adding the attributes name,price,category
	def __init__(self,name,price,category):
		self.name = name
		self.price = price
		self.category = category

	# creating the Method get_info()
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")

	# Extra(Optional) :- adding a Method apply_discount(percent) that return the discounted price

	def apply_discount(self,percent):
		discount = (self.price * percent) / 100
		discounted_price = self.price - discount
		print(f"The Price after {percent}% discount is {discounted_price} ")

```

```python
pro = Product("Laptop" , 55000 , "Electronic")
```

```python
pro1 = Product("Mouse" , 1000 , "Accessories")
```

```python
pro.get_info()
pro.apply_discount(10)
```

```python
pro1.get_info()
pro1.apply_discount(5)
```
