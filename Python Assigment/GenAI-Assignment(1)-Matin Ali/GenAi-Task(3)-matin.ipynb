# 1. Create a dictionary_price where keys are product names and values are price.include at least 6 entries

```python
# creating a dictionary price_dict with keys are product name and values are price

dict_price = {"Laptop" : 750000,
	      "Headphone" : 15000,
	      "T.V" : 25000,
	      "Mouse" : 1500,
	      "Monitor" : 2000,
	      "webcam" : 3000,
	      }
```

# 2. Write small Code block to

```python
# 2. Add new product with price to price_dict

dict_price["Printer"] = 15000 # adding the new product with price
dict_price
```

```python
# 2.1 Update the price of existing product

dict_price["Mouse"] = 2000 #updating the price of existing product mouse
dict_price
```

```python
# 2.2 remove the product by name 
pr = input("enter the name of the product to remove")
if pr in dict_price: # checking the product is exist or not
    del dict_price[pr]
    print(f"{pr} this product is removed ")
else:
    print(f"{pr} is not existing product")

dict_price
```

# 3 Print the average price of all products 

```python
# calculating the average price 
tp = sum(dict_price.values())
np = len(dict_price)

avg = round(tp / np , 2) # average price calculating
print(f"The average price of all the product is : \n {avg}")
```

# Extra (Opational) :- Print the product with the both maximum and minimum prices

```python
# printing the maximum product price
mp = max(dict_price, key=dict_price.get) 
print(f"Maximum price product is : {mp}")

# printing the minimum product price
minp = min(dict_price,key=dict_price.get)
print(f"Minimum product price is : {minp}")
```
