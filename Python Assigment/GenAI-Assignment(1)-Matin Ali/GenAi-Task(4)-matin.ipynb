# 1. Using the product list and price_dict, create a list of tuple name catalog where each tuple is (product_name,price,category)

```python
# Creating the product list
products = ["Laptop",
	   "Mouse",
	   "T.V",
	   "Keyboard",
	   "Headphone",
	   "webcam"]
products
```

```python
# creating a price_dict
price_dict = {"Laptop" : 750000,
	      "Headphone" : 15000,
	      "T.V" : 25000,
	      "Mouse" : 1500,
	      "Keyboard" : 2000,
	      "webcam" : 3000,
	      }
price_dict
```

```python
# creating list of category
category = ["Electronic",
	    "Audio",
	    "Electronic",
	    "Electronic",
	    "Accessories",
	    "Electronic"]
category
```

```python
# Creating a list of tuples catalog
catalog = []

for i in range(len(products)):
	catalog.append((products[i],
	price_dict[products[i]],
category[i]))

catalog
```

# 2. From catalog, create a new dictionary category_to_product maps each category to a list of product name in that category

```python
category_to_products = {}
for product,price,category in catalog:
	if category not in category_to_products:
		category_to_products[category] = []
	category_to_products[category].append(product)

category_to_products
```

# 3. Print all product that belong to the category that has the maximum number of products

```python
max_category = max(category_to_products, key= lambda x : len(category_to_products[x]))

print(f"Category : {max_category}")
for product in category_to_products[max_category]:
	print(product)
```
