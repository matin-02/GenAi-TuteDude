# Task 2 : Read File in Different Ways

```python
# Reading the entire file using .read() and print it

with open('sales_data.txt','r') as file:
	content = file.read()
	print(f"Using The .read(): {content}")
```

```python
# reading the first line using .readline()

with open('sales_data.txt','r') as file:
    content = file.readline()
    print(f"Read the first line using .readline() : {content}")
```

```python
# Reading the all lines Using the .readlines()

with open('sales_data.txt','r') as file:
     lines = file.readlines()
     sales = []
     for line in lines:
         sales.append(int(line.strip()))
     print(f"Reading the all lines using .redalines() {sales}")
```
