# Task 5 :- Mini Program : Safe Shopping Cart

```python
# 1. has a cart list 
cart = []

# 2. Runs a loop asking user to enter prices.
while True:
    price = input("Enter Product Price (or 'q' to quit)")
    # 3. stops when user enter 'q'
    if price == 'q':
        break
    try:
        # 4.1 convert input to float
        price = float(price)

	# raise custom exception if price is negative
        if price < 0:
            raise ValueError("Negative Price Not Allowed")
        
        cart.append(price)

	# Handling ValueError if user enter invalid input
    except ValueError as e:
        print("Invalid Price : ",e)

# 5. at the end,print total items and total bill 
# additoinlly items in cart for better understanding of cart storing the values

print('Total Items : ' , len(cart))
print('Total Bill : ' , sum(cart))
print("The items in cart is" , cart)
```
