# Task 2 :- Bill Calculator With error Handling

```python
# Given list of product prices:
# prices = [120, 350, 'abc',500,-200,800]
# Write code that
# 1. iterate through the list
prices = [120, 350, 'abc',500,-200,800]
total = 0

for price in prices:
	# 2. tries to add only valid(positive numerical) prices to the total
	# 3.Handles:
	# TypeError
	# using raise custom exception ValueError
	try : 
		if price < 0:
			raise ValueError(f'Negative Price Not Allowed {price}')

		# printing the running total
		# skipping the invalid items but continue processing
		total += price
		print(f"The running total is : {total} after adding the {price}")

	# TypeError Handling
	except TypeError:
		print(f"Only Numerical values : '{price}' is not Numeric Value")

	# ValueError handling with custom exception
	except ValueError as e:
		print(e)
# printing the final total
	
print(f"final total : {total}")
```
