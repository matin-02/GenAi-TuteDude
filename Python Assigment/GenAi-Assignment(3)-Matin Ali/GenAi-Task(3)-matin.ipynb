# Task 3 :- Lambda Function GST Calculator

```python
# creating a lambda function gst to return price after adding 18% GST

# creating a lambda gst function 
gst = lambda price: price + (price * 0.18)
```

```python
gst(100)
```

```python
gst(1000)
```

```python
# Creatinga aanother lambda to compute final price afetr GST + discount together
# calculating the final price after applying the GST + discount

final_price = lambda price,discount: gst(price) - (gst(price) * (discount/100))
```

```python
final_price(100,5)
```
