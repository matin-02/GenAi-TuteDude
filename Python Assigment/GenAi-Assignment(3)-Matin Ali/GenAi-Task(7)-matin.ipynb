```python
# creating a three function to add the price andto calculate the average price and to get the maximum price from the prices list

# creating a function add_price

def add_price(prices_list,price):
	prices_list.append(price)

# creating a function get_average_price(price_list)
def get_average_price(prices_list):

#calculating and returning the average price
	return sum(prices_list) / len(prices_list)

# creating a function get_max_price(prices_list)
def get_max_price(prices_list):

# returning the maximum price from the price list
	return max(prices_list)
```
