# Task 2 Recursive Function Factorial 


```python
# create a recursive function factorial(n)

# creating a function factorial

def factorial(n):

# Prints an error message if the n is negative number
	if n < 0:
		print("Error : Factorial is not defined for negative numbers")
		return

# handling the case when n is 0 or 1 
	if n == 0 or n == 1:
		return 1 

# Returning the factorial of n		
	return n * factorial(n - 1)
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```
