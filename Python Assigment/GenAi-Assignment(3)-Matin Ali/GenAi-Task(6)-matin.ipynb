# Task 6 :- Combined Utility Function

```python
# Creating a function process Price 
def process_price(prices):
	# using a map() + lambda to apply 10% disount to all the prices
	discounted_prices = list(map(lambda price : price - (price * 0.10), prices))

	# using a filter() to keep the only discounted prices that are above 300
	filtered_prices = list(filter(lambda price : price > 300, discounted_prices))

	# returning both the lists
	return discounted_prices, filtered_prices
```

```python
# creating a list of prices
discounted_prices, filtered_prices = process_price([100 , 500 , 500 , 900 , 50 , 750])

# printing both the lists
print("Discounted Prices : ", discounted_prices)
print("Filtered Prices : ", filtered_prices)
```
