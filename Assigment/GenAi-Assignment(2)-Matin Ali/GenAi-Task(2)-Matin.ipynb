# 1. Given a list order amounts , use a for loop to apply the discount rule 

```python
# 1.creating the order amountts list
orders = [1200,2500,800,1750,3000]

# applying the discount rules
for orders_amount in orders:
	if orders_amount >= 2000:
		discount = 15
	elif 1500 <= orders_amount < 2000:
		discount = 10
	elif 1000 <= orders_amount < 1500:
		discount = 7
	else:
		discount = 0
	
	final_amount = orders_amount - (orders_amount * discount / 100)

	print(f"Orders_amount : {orders_amount}  -->  Discount : {discount}%  -->  Final_Amount : {final_amount}")
```

```python
# 2. Computing and printing the total revenue after discount
orders = [1200,2500,800,1750,3000]
# variable to track the revenue
total_revenue = 0

# applying the discount rules
for orders_amount in orders:
	if orders_amount >= 2000:
		discount = 15
	elif 1500 <= orders_amount < 2000:
		discount = 10
	elif 1000 <= orders_amount < 1500:
		discount = 7
	else:
		discount = 0
	
	final_amount = orders_amount - (orders_amount * discount / 100)

# Calculation the total revenue values 
	total_revenue = total_revenue +  final_amount

	print(f"Orders_amount : {orders_amount}  -->  Discount : {discount}%  -->  Final_Amount : {final_amount}")
	
print("-" * 100)
	
print(f"The Total Revenue after the Discount is : {total_revenue}")
```
