1. given a list of daily sales iterate through it with a for loop and : 
  - if day's value is -1 : treat it as corrupted data and break the loop
  - if day's value is 0 : treat it as a day with no sales and continue
  - for valid positive sales , add to total_sales and print the running total

```python
# creatin a list of daily sales 
daily = [200 , 150 , 0 , 400 , 50 , -1 , 300]
total_sales = 0
# itreating through it with for loop
for sales in daily:
	# checking the sales is -1 or what if -1 Breaking the loop
	if sales == -1:
		break

	# checkin the sales is 0 or what if 0 then skpping the adding to revenue
	if sales == 0:
		continue

	total_sales += sales
	print(f"Running Total : {sales}")
```
