# Task 5 :- Create Product Info File(User Input)

```python
# 1. Ask the User for 3 product names and their prices
products = []
print("Please Enter the 3 Product and their Prices : ")

for i in range(3):
	name = input(f"Enter a name of the product {i + 1} : ").strip()
	price = int(input(f"Enter the price for {name} : "))
	products.append((name , price))
```

```python
# 2 & 3. Write them into a new file products.txt in this format productName | Price
with open('products.txt','w') as file:
    for name,price in products:
        file.write(f"{name}  |  {price} \n")
```

```python
# Read the file and print each line proper formatting
with open('products.txt','r') as file:
    content = file.readlines()
    for lines in content:
        print(lines)
```
