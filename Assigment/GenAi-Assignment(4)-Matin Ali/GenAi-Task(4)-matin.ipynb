# Task 4 :- Generate Summary Report from file
#### Using Only file reading operation

```python
# 1. read all the sale values from sales_data.txt

with open('sales_data.txt','r') as file:
	lines = file.readlines()
```

```python
# 2. convert them into integer
sales = []
for line in lines:
	sales.append(int(line.strip()))
```

```python
# 3 .Calculate and print Total sales, highest sale, lowest sale,average sale
total_sale = sum(sales)
highest_sale = max(sales)
lowest_sale = min(sales)
average_sale = total_sale / len(sales)

# printing the summary
print(f"Total sales  is : {total_sale}")
print(f"Highest sale is : {highest_sale}")
print(f"Lowest sale is : {lowest_sale}")
print(f"Average of sale is : {average_sale}")
```
