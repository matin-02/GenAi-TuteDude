# Task 1 : Write sales Records to a file

```python
# Creating a list of sales amount
sales = [1200,450,980,1500,3000]

# opening the file sale_data in write mode
file = open('sales_data.txt','w')

# writing the each sale on the new line
for sale in sales:
	file.write(f"{sale}\n")
	
# closing the file
file.close()
```

```python
# reopening the file again
files = open('sales_data.txt','r')
content = files.read()
print(content)
files.close()
```

```python
# Extra (optional) : creating and writing on extra on different programm
# for the better understanding of the code and program execution

sales_str = [str(sale) for sale in sales]

# joining the them with commas

comma_sep = " , ".join(sales_str)

files = open('sales_data.txt','w')
files.write(comma_sep)
files.close()
```
