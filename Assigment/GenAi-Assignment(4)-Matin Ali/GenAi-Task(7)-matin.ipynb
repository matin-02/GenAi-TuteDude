# Task 7 :- Mini Project - Export Discounted Prices
 

```python
# create a dictionary

prices = {
	"Mouse" : 500,
	"Keyboard" : 800,
	"Monitor" : 7000,
	"Pendrive" : 400,
	"Camera" : 5000
	}
```

```python
# Ask the user for a discount percenatge

discount = int(input("Enter a Discount Price"))
```

```python
# write discounted proce to the file
discounted_price = []
with open('discount_report.txt', 'w') as file:
    file.write('Product | Original Price | Discounted Price\n')
    for product,price in prices.items():
        final_price = price - (price * discount / 100)
        discounted_price.append(final_price)
        file.write(f"{product}  |  {price}  |  {final_price}\n")
```

```python
# After writing , read the file and print it to the terminal
with open('discount_report.txt','r') as file:
    content = file.read()
    print(content)
```

```python
# Optinal(Extra) :- Write Summary at the bottom of the file total items and average discounted price
total_items = len(prices)
average_price = sum(discounted_price) / total_items

with open('discount_report.txt', 'a') as file:
    file.write("\n")
    file.write(f"Total Items : {total_items}\n")
    file.write(f"Average Discounted Price is : {average_price}")
    
```
