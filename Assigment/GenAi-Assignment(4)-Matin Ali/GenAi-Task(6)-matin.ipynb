```python
# 1. Ask the user for a filename to open.
import os
filename = input("Enter a File name with proper file format : ")
```

```python
# 2. if the file exists, read and print it
if os.path.exists(filename):
	with open(filename,'r') as file:
		print(file.read())
else:
	print(f"File Not Found {filename}. Please Check the filename")
```
