# Task5 : Using filter() : filter Expensive products

```python
# Given the list of prices

prices = [100 , 250 , 400 , 1200 , 50 , 2000 , 850]

print(f"Original Prices : {prices}")

# using filter() to create list of price greater than 500

price_greater = list(filter(lambda price: price > 500, prices))

print(f"Prices Greater than 500 : {price_greater}")

# using filter() to create list of price less than 500

price_less = list(filter(lambda price: price < 500, prices))

print(f"Prices Less than 500 : {price_less}")
```
