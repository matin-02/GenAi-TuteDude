# Task 4 Using map() : apply GST to List of prices

```python
# use a map() with the gst lambda to generate a new list prices_with_gst

prices = [100 , 250 , 400 , 1200 , 50]

# creating the new list price_with_gst using a map() with gst lambda function
prices_with_gst = list(map(lambda price: price + (price * 0.18),prices))

# printing the orginal prices list
print(f"original prices : {prices}")

# printing the new prices_with_gst list
print(f"Price after GST : {prices_with_gst}")
```
