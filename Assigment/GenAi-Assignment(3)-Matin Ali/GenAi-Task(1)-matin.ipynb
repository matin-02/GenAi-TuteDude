# Task 1 :- Basic Function Price after Discount

```python
# 1. write a Function apply(price,discount_percent)
# creating a function apply_discount and calculating the price and returnin the final price
def apply_discount(price, discount_percent):

# calculating the discount and final price
	discount = price * discount_percent / 100
	final_price = price - discount

# returning the final price after applying the discount
	return final_price
```

```python
f = apply_discount(1000, 10)
print(f"The final Price after the discount is : {f}")
```

```python
# 2. if the discount is missing then apply default discount of 5
# creating a function apply_discount and calculating the price and returnin the final price
# applying the default didcount of 5% if the discount is missing

def apply_discount_default(price, discount_percent = 5):# default discount of 5%

# calculating the discount and final price
	discount = price * discount_percent / 100
	final_price = price - discount

# returning the final price after applying the discount
	return final_price
```

```python
f2 = apply_discount_default(500)
print(f"The final price after the default discount is : {f2}")
```

```python
print(apply_discount_default(1000,10))
print(apply_discount_default(500))
```

```python
# Extra(Optionsal) : add a condition to ensure discount never exceeds 60%
#creating a function apply_discount and calculating the price and returnin the final price
# applying the default didcount of 5% if the discount is missing

def apply_discount_optional(price, discount_percent = 5):# default discount of 5%

# adding the condition to ensure discount never exceeds 60%
	if discount_percent > 60:
		discount_percent = 60

# calculating the discount and final price
	discount = price * discount_percent / 100
	final_price = price - discount

# returning the final price after applying the discount
	return final_price
```

```python
print(apply_discount_optional(1000,10))
print(apply_discount_optional(500))
print(apply_discount_optional(1000,80)) # discount exceeds 60% so it will be capped at 60%
```
