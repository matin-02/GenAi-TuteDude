# Task 1 :- Safe Division Utility

```python
# Write a program that :
# 1 . Take two inputs from the user : numerator and denominator.

# 2.try - except to handle ValueError and ZeroDivisionError 
try:
	numerator = int(input("Enter a Numerator Values :"))
	denominator = int(input("Enter a Denominator Value : "))

	result =  numerator / denominator

except ValueError as ex:
	print(ex)

except ZeroDivisionError as ex1:
	print(ex1)

# 3 . if no error occurs, print the result inside the else block 
else:
	print(f"The Division of the given Numerator {numerator} and Denominator {denominator} is : {result}")

# 4. in finally block, print:"Operation Completed"
finally:
	print("Operation Complete")
```
