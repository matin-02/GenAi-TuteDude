```python
# Creating a class Product
class Product:
	# adding the attributes name,price,category
	def __init__(self,name,price,category):
		self.name = name
		self.price = price
		self.category = category

	# creating the Method get_info()
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")


	# Adding the  __str__ methods 
	def __str__(self):
		# returning a readble string Product(name,price,category)
		return f"Product ( {self.name} , {self.price} , {self.category})"

	# Operator Loading with(__add__) method
	def __add__(self,other):
		# allows product1 + product2 
		# to return the total combined price.
		return self.price + other.price


# creating a class Inventory with attributes list[]
class Inventory:
	def __init__(self):
		self.products = []

	# adding the product using add_product method
	def add_product(self,product):
		self.products.append(product)

	# creating a remove_products method to remove the product by name
	def remove_products(self, name):
		for product in self.products:
			if product.name == name:
				self.products.remove(products)
				print(f"{name} Removed from Inventory")
				return
		print(name , "Not Found")

		# creating a get_total_value method to get the sum prices of all products
	def	get_total_value(self):
		total = 0
		for product in self.products:
			total += product.price

		return total

		# creating a show_all_products method to print the info each product
	def show_all_products(self):
		for product in self.products:
			product.get_info()
			print()

# creating a class Store

class Store:
	def __init__(self,store_name):
		self.store_name = store_name
		self.inventory = Inventory()

	def add_new_product(self):
		name = input("Enter a product Name : ")
		price = int(input("Enter a price of product : "))
		category = input("Enter product category : ")

		product = Product(name , price , category)
		self.inventory.add_product(product)

	def show_summary(self):
		print("Store Name : " , self.store_name)
		print("Total Items : " , len(self.inventory.products))
		print("Total Value : " , self.inventory.get_total_value())

```
