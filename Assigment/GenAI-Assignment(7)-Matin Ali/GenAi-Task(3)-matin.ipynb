# Inheritance (Single - Level)

```python
class Product:
	# adding the attributes name,price,category
	def __init__(self,name,price,category):
		self.name = name
		self.price = price
		self.category = category

	# creating the Method get_info()
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")

# creating the subclass ElectrionicProduct that inherits from product.
class ElectronicProducts(Product):
	def __init__(self,name,price,category,warranty_years):
		super().__init__(name,price,category)
		self.warranty_years = warranty_years

	# Overriding the get_info() method in ElectronicProduct
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category} \nWarranty Years : {self.warranty_years}")
```

```python
pro = Product("Laptop" , 55000 , "Electronic")
```

```python
pro.get_info()
```

```python
# creating an object and inheriting the product class name,price,and category attributes and adding the warranty_years
pro = ElectronicProducts("Laptop" , 55000 , "Electronic",2)
```

```python
pro.get_info()
```
