# Task 2:- Constructor and Encapsulation

```python
# Modifying the Product class Making the price attribute private

class Product:
	def __init__(self,name,price,category):
		self.name = name
		self.__price = price # making the price attribute private by using __
		self.category = category

	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")

	# creating a getter and setter methods:
	# get_price method to get the private attribute price to print
	def get_price(self):
		print(f"The Original Price of {self.name} is {self.__price}")

	# set_price methods to set the new price of the product
	def set_price(self,new_price):
		if new_price > 0: # checking the price must be greater than 0
			self.__price = new_price # updating the new price of product
			print(f"The Updated price of {self.name} is {new_price}")
		else:
			print(f"The price must be Greater Than 0 You Entered {new_price}")
```

```python
pro = Product("Laptop" , 55000 , "Electronic")
```

```python
pro1 = Product("Mouse" , 1000 , "Accessories")
```

```python
pro.get_info()
# the price is not printing because the price is private there
```
