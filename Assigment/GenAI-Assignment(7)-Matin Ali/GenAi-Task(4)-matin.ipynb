# Task 4:- Polymorphism

```python
class Product:
	# adding the attributes name,price,category
	def __init__(self,name,price,category):
		self.name = name
		self.price = price
		self.category = category

	# creating the Method get_info()
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")

# creating a 1st class laptop
# both the class overriding the get_info method
class Laptop(Product):
	def get_info(self):
		print(f"Name of Laptop is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")

# creating 2nd class Mobile
class Mobile(Product):
	def get_info(self):
		print(f"Name of Mobile is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")
```

```python
laptop = Laptop("Acer" , 55000 , "Laptops")
```

```python
mobile = Mobile("Vivo" , 45000 , "Mobiles")
```

```python
# Writing a loop to iterate over object laptop and mobile and calling the get_info() product
products = [laptop,mobile]

# iterating and printing the products info using get_info() method
for product in products:
    product.get_info()
    print()
```
