# Task 6 :- Magic Method And Operator Overloading

```python
# Creating a class Product
class Product:
	# adding the attributes name,price,category
	def __init__(self,name,price,category):
		self.name = name
		self.price = price
		self.category = category

	# creating the Method get_info()
	def get_info(self):
		print(f"Name of Product is : {self.name} \nPrice is : {self.price} \nProduct Category : {self.category}")


	# Adding the  __str__ methods 
	def __str__(self):
		# returning a readble string Product(name,price,category)
		return f"Product ( {self.name} , {self.price} , {self.category})"

	# Operator Loading with(__add__) method
	def __add__(self,other):
		# allows product1 + product2 
		# to return the total combined price.
		return self.price + other.price
```

```python
# Testing this with two product objects
product1 = Product("Acer Laptop" , 55000 , "Electronis")
```

```python
print(product1)
```

```python
product2 = Product("Gaming Mouse" , 1500 , "Accessories")
```

```python
print(product2)
```

```python
print(product1)
print(product2)
```

```python
total_price = product1 + product2
```

```python
print(f"The Total Combined Price is : {total_price}")
```
