# Task 5 :- Abstraction (Using Abstract Base Class)

```python
# import the abstract module and method
from abc import ABC,abstractmethod 

# creating the abstract class Payment 
class Payment(ABC):
	# creating abstract method process_payment
	@abstractmethod
	def process_payment(self,amount):
		pass

# creating a class CreditCardPayment
class CreditCardPayment(Payment):
	# overriding the process_payment method with simple print statment 
	def process_payment(self, amount):
		print(f"Credit Card Payment Proceed : {amount}")

# creating a class UPIPayment
class UPIPayment(Payment):
	# overriding the process_payment method with simple print statment
	def process_payment(self, amount):
		print(f"UPI Payment Proceed : {amount}")
```

```python
# testing all the test cases
# creating a object of CreditCardPayment
ccp = CreditCardPayment()
```

```python
ccp.process_payment(5000)
```

```python
upi = UPIPayment()
```

```python
upi.process_payment(1000)
```
