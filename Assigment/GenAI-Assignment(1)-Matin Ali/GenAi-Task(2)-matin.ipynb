# 1. Creating the Categories set from the product list (creating short prallel list of categories)

```python
category = ["Electronic",
	    "Electronic",
	    "Electronic",
	    "Electronic",
	    "Electronic",
	    "Audio"]

```

```python
# Creating the unique set of category
category_Set = set(category)
```

```python
category_Set
```

# 2. Adding the new Category to the set and showing that duplicates are ignored

```python
category_Set.add("Audio") #there we are adding the duplicate category in set
print(f"The Duplicate Category is get ignored in this set as the Audio is already Present in the set : \n {category_Set} ")
```

```python
category_Set.add("Office") # there we are adding the new category in set
print(f"The new value get updated in set as it is a unique value in the set :\n {category_Set}")
```

# 3. Checking Weather a Category is exists in the set or not (printing the boolean value)

```python
print(f"Checking the Office is present in the category or not : \n {"Office" in category_Set}") # it will return true as the office is present in the set
print()

print(f"Checking the Perfume is present in the category or not : \n {"Perfume" in category_Set}") # it will return false as the Perfume is not present in set
```
