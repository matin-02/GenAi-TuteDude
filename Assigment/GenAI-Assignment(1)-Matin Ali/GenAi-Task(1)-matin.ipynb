Creating the product list with 6 products names with String data structure 

```python
products = ["laptop", "mobile", "tablet", "desktop", "smartwatch","headphones"] # creating a list of products
print(products) # displaying the list
```

Creating a tuple sample_product with (product name, price,category) for on product

```python
sample_products = ("laptop",65000,"Electronics") # creating a tuple of products
print(sample_products) # displaying the tuple
```

print 2nd and last product from the product list

```python
print("The Second product is : ", products[1])  # Accessing the second product

print("The Last product is : ", products[-1])  # Accessing the last product
```

Appending the new product name to the products list and printing the updated products list

```python
products.append("camera")  # Adding a new product to the list
print("Updated products list: ", products)  # Displaying the updated list
```

Optional Task to convert the sample_product tuple to list and update the price and covert the list into tuple


```python
sp_li = list(sample_products)  # Converting the tuple to a list
print("Converted list from tuple: ", sp_li)  # Displaying the converted list

sp_li[1] = 70000  # Modifying the price in the list
print("Modified list: ", sp_li)  # Displaying the modified list

sp_tu = tuple(sp_li)  # Converting the list back to a tuple
print("Converted tuple from list: ", sp_tu)  # Displaying the converted tuple
```
