## CSV Library --> Most of the time we are going to work with csv like that why knowledge with the csv library functionalities matters

```python
import csv
```

```python
with open('example.txt',mode = 'w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Amit',25])
```

```python
with open('example.txt', mode = 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Json Library

### When we are going to work with NoSql Database data then the data inside the those NoSQL database is going to ne their in the Json format and along with that when we are going to work with the dataset  as well then again these dataset will be consisting of data in the json format itself

```python
# when we get the Json Data, we get in a Json String Format so by using json library we can convert 
# our string data into dictionary farmat for serializing our data
```

```python
import json
```

```python
data = {"Name" : "matin","Age" : 25}
str_json = json.dumps(data)
print(str_json)
type(str_json)
```

```python
dict_json = json.loads(str_json)
print(dict_json)
type(dict_json)
```

## Regular Expression Library

## ragular expression is used to retrive info based on a specific pattern or if the pattern matches

```python
text = 'hello i name is matin my password is hellomatin12345'
```
