# Pillars of Object Oriented Programming

## Inheritance :-
#### Inheritence is a Fundamental concept in Object Oriented Programming because og whivh the child class is able to inherit the proerties Or Attributes and functionalities of the parent class 

#### Just like a child is going to inherit the propertoes from parent and parent is going to inherit the properties from the grandparent and definitiely in this hierarchy the grandchild can have some features different from the parent and grandparent as well


## In Python typically we are going to see 3 types of Inheritance
### 1. Single Inheritance
### 2. Multilevel Inheritance
### 3. Multiple Inheriatance


### Single Inheritance :- The features or the Attributes from the parent class will be getting inherited into one child class.

### |
### Chlid Class

## Multilevel Inheriatnce ---> The porperties or Methods and Attributes of the Grandparent class can be inherited by the parent class and then the properties of the parent class will be inherited by the child class .In a proper heirarchy the Inheritance process going to work for us.

## Multiple Inheritance :- In Multiple Inheritance the properties of two parent class or multiple parent class will be getting inhertied into the child class

### Parent (Properties) + Parent_1 (Properties) + Parent_ 2 (Properties) + Parent_3 (Properties) + Parent_4 (Properties)...
### |
### |
### Child

```python
class Avenger:
    def __init__(self,name,power):
        self.name = name
        self.power = power

    def intro(self):
        print(f"Hello My Name is {self.name}")

class Ironman(Avenger):
    def __init__(self,name,power,weapon):
        super().__init__(name , power)
        self.weapon = weapon

    def weapon_details(self):
        print(f"I use {self.weapon}")
```

```python
iron = Ironman("IronMan" , "Genius Intellect" , "Repulser beams")
```

# Mutilevel Inheritance

```python
class Grandparent:
    
    def __init__(self,nationality ,baldness , anger_issues):
        self.nationality = nationality
        self.baldness = baldness
        self.anger_issues = anger_issues

    def intro(self):
        print(f"Hello My Nationality is : {self.nationality} and about my anger issues if you want to ask about then the answer is : {self.anger_issues}")
        
```

```python
class Parent(Grandparent):
	def __init__(self, nationality, baldness, anger_issues, intelligence):
		super().__init__(nationality, baldness, anger_issues)
		self.intelligence = intelligence

	def intelligence_details(self):
		print(f"About My intelligence  if you wnat to know that wheather if i am intelligent or not :  {self.intelligence}")
        
```

```python
class Child(Parent):
    def __init__(self,nationality,baldness,anger_issues,intelligence,charm):
        super().__init__(nationality,baldness,anger_issues,intelligence)
        self.charm = charm

    def char_details(self):
        print(f"About My  Charm : {self.charm}")
```

```python
p1 = Parent("Indain" , "Yes" , "No","Yes")
```

```python
p1.intro()
```

```python
ch1 = Child("Indian" , "No" , "Yes" , "Yes" , "Yes")
```

```python
ch1.charm
```

```python
ch1.intro()
```

```python
ch1.intelligence_details()
```

```python
ch1.char_details()
```

```python
ch1.nationality = "American"
```

```python
ch1.intro()
```

```python
class Ironman:
	def __init__(self , name , power , suit , weapon , nationality = "American"):
		self.name = name
		self.power = power
		self.suit = suit
		self.weapon = weapon
		self.nationality = nationality

	def intro(self):
		print(f"My name is :  {self.name} and powers are :{self.power}")  


class CaptainAmerica:
	def __init__(self,name,power,suit,weapon,nationality = "American"):
		self.name = name
		self.power = power
		self.weapon = weapon
		self.nationality = nationality

	def weapon_details(self):
			print(f"I cary the weapon as : {self.weapon}")
```

```python
class Hulk(Ironman,CaptainAmerica):

	def __init__(self,name,power,suit,weapon,anger_issues ,nationality = "American"):
		super().__init__(name,power,suit,weapon,nationality)
		self.anger_issues = anger_issues

	def anger_desc(self):
		print(f"About My Anger Issues I have or not : {self.anger_issues}")
```
