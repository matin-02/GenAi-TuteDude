## Types Of Methods :

### 1. Instance Methods :-  Instance methods operate on Object Attributes . Instance Method is going to use self keyword to access and modify attributes . This Methods are called using an objects.

### 2. Class Methods :- It is the method of the whole class or the blueprint which will remain the same for all of the object then such kind of Method is being reffered as a class Method . In class Method , these method work with the class level . They are going to use "cls" instead of self . they are called using the class name . for specifiying that the current method is a class method we have to use the abbreviation as  @classmethod

### 3. Static Methods :- Static Methods do not use self or cls . They work like normal functions but belong to a class. they are going to be designed by the abbrevation @staticmethod . static method can be also be called by the classname also

```python
class Avengers:
    def __init__(self , name , power):
        self.name = name
        self.power = power

    def show_details(self):
            print(f"My name is {self.name} and i am having power {self.power}")
```

```python
iron_man = Avengers("Iron Man" , "Genius Intellect")
```

```python
iron_man.show_details()
```

```python
thor = Avengers("Thor" , "God Of Thunder")
```

```python
thor.show_details()
```

```python
class Avengerss:

	origin = "Avengers"
	def __init__(self,name,power):
		self.name = name
		self.power = power

	@classmethod
	def origin_info(cls):
		print(f"My origin is {cls.origin}")

	def self_info(self):
		print(f"My name is {self.name} and my power is {self.power}")
```

```python
iron_man = Avengerss("Iron Man" , "Genuis Intellect")
```

```python
iron_man.power
```

```python
iron_man.self_info()
```

```python
iron_man.origin_info()
```

```python
thor1 = Avengerss("Thor" , "God Of Thunders")
```

```python
thor1.self_info()
```

```python
thor1.origin_info()
```
