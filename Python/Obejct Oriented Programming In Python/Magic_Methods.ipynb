# Magic Methods

## Magic Methods are also called as Dundar Methods because they have double underscore __ befor and after their name and they are Special methods in python that allow us to customize how obejcts behave .

## These Methods can hepls us define :
## 1. object Inialization(__init__)
## 2. String Representation(__str__,__repr__)
## 3. Arthimetic Operations(__add__, ___sub__,etc)
## 4. Comparison Operations(__eq__,__lt__)
## 5. Length, indexing,iteration(__len__,__getitem__,etc)

## Why to use Magic Methods

### 1. Make objects behave like built in type(eg , lists , integer)
### 2. Improves Readability by allowing basic operation object1 + object2

### 3. Simplifies code by replacing verbose function calls with natural syntax

## print(object) --> __str__ --> what is returned when print(object) is used



### Arithmetic operations can be done among the objects by using Magic Methods

```python
## __add__

class Avenger:
	def __init__(self,name,strength):
		self.name = name
		self.strength = strength

	def __add__(self,other):
		return Avenger(self.name +  " & " + other.name , self.strength + other.strength)

	def __str__(self):
		return f"{self.name} with the combined strength {self.strength}"

ironman = Avenger("Iron Man" , 80)
thor = Avenger("Thor" , 100)
team = ironman + thor
print(team)
```
