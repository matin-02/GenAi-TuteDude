# Encapsulation

## Encapsulation is a way of wrapping data (variables and Methods) inside a capsule or a class as a single unit .

## It restrict the direct access to some of the objects components which is a means of preventing accidental inference and misuse of the data .

## Key Concepts of Encapsulation

### 1. Public Members
### 2. Protected Members
### 2. Private Members
### getter and setter methods

## Public Member :-
### Accessible from Anywhere or Attributes are accessible outside of the class and These members are accessible by the derived class
### public member attributes is directly declared

```python
class Ironman:
	def __init__(self,name,power,weapon):
		self.name = name
		self.power = power
		self.weapon = weapon

	def intro(self):
		print(f"Hello My Name is {self.name} and I have {self.weapon}")
```

```python
## dir() --> it takes the input as the objects name and gives us idea that which member is public or protected or private
```

```python
iron = Ironman("Iron Man" , "Genius intellect" , "Repulser Beam")
```

```python
dir(iron)
```

```python
iron.name
```

```python
iron.intro()
```

```python
iron.weapon
```

## Private Member :- 
### A private Member is not accessible outside the class and not even in the derived class and in order to have the access over these private member or to bring some modification on these members use getter and setter method .

### Private Members are decalred by using double underscore __ before the name

```python
class Avengers:
    def __init__(self, name , phone_no ):
        self.name = name
        self.__phone_no = phone_no
```

```python
aveng = Avengers("captain" , 74854125)
```

```python
aveng.name
```

```python
aveng.phone_no
```

```python
dir(aveng)
```

## getter Function

### getter Method ---> it will be helping us out with accessing the private data

```python
class Captain:
    def __init__(self,name,address):
        self.name = name
        self.__address = address

    ## getter method
    def accessing_add(self):
        print(f"My Address is {self.__address}")
```

```python
captain = Captain("Captain America" , "New York")
```

```python
captain.name
```

```python
captain.address
```

```python
dir(captain)
```

```python
captain.__address
```

```python
captain.accessing_add()
```

## setter method 
### this being used to mofifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name = name
        self.__sound = sound

    def sound_info(self): ## setter method
        print(f"the animal is going to sound as : {self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound = new_sound
        print(f"New sound for the animal is : {self.__sound}")
```

```python
animal = Animal("Bruno" , "Woof Woof")
```

```python
animal.name
```

```python
animal.sound
```

```python
### getter method to have the access over this data
```

```python
animal.sound_info()
```

```python
animal.sound_modify("meow meow")
```

## Protected Members : -
### Protected data is the data which is not accessible outside the class but the protected data is going to be accessible by the derived class or child class.
### For having the protected member you can mention your protected member as attributee or by using single _ underscore

```python
class Grandparent:
    def __init__(self,name,locker_key):
        self.name = name
        self._locker_key = locker_key

    # getter method
    def locker_info(self):
        print(f"The locker key info is : {self._locker_key}")
```

```python
class Parent(Grandparent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveal(self):
        print(f"the Locker key information is : {self._locker_key}")
```

```python
grandpa = Grandparent("abc", 12450)
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
grandpa.locker_info()
```

```python
dir(grandpa)
```

```python
parent = Parent("ABX" , 124501)
```

```python
parent.locker_key
```

```python
parent.name
```

```python
parent.locker_info()
```

```python
parent.child_reveal()
```

```python
## Bank Account System
```

```python
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self):
        print(f"The Total Balance is : {self.__balance}")


    def set_balance(self,amount):
        if amount > 0:
            self.__balance = amount

            return f"Deposites Amount is : {self.get_balance()}"
        
        else:
            return f"The Deposited Amount must be Positive"

    def withdraw(self,amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            return f" Withdraw {amount} is : {self.get_balance()}"

        else:
            return "Insufficient funds in the account"
```

```python
account = BankAccount("John" , 100000)
```

```python
account.account_holder
```

```python
account.balance
```

```python
account.get_balance()
```

```python
account.withdraw(5000)
```

```python
account.get_balance()
```
