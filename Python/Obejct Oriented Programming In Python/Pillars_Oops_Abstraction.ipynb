# Abstraction

## Abstraction is the concept og hiding the implementation details and only showing the necessary features of the object.

## This helps in reducing programming complexity and efforts

## Abstraction allows us to work with an object with what an object does rather than how it does it 

## How can we achieve abstraction in python

### 1. Abstract class ---> A class That Cannot be instantiated or that means for an abstract class we cant create the object of this class

### 2. Abstract Method ---> Abstract method just because it cant accessed by the obejct of the abstract class itself thats why we have to call this abstract method in our child class and the implemenation of the abstract method is not provided inside the abstract class but must be provided in the child class

### The abstract method is being defined by @abstractmethod

```python
from abc import ABC,abstractmethod

class Vechile(ABC):
    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vechile):
     def start(self):
           print("The car starts with an ignition start key")
     def stop(self):
          print("The car stops")

class ElectricScooter(Vechile):
     def start(self):
          print("Electric Scooter start with power button")
     def stop(self):
          print("Electric Scooter Stop with an electric brake")   
```

```python
car = Car()
```

```python
car.start()
```

```python
car.stop()
```

```python
es = ElectricScooter()
```

```python
es.start()
```

```python
es.stop()
```
