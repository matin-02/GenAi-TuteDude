## In the OOPs polymorphism allows object of different classes as if they are the obejct of the same class and It enables reusability and make programs more flexible
 

## Types of Polymorphism:
### 1. Method Overriding (Runtime Polymorphism)
### 2. Mehtod Overloading (complie Time ploymorphism)
### 3. operator Overloading 

### Method Overriding ---> A child Class provides a specific implementation of a method that has already been defined into the parent class
### This allows Dynamic selection at runtime

```python
class Grandparent: # Parent Class

	def speak(self):
		print("Hello My name is Matin")

class Parent(Grandparent):
	def speak(self):
		print("Hello My name is Junior Matin")

class Child(Grandparent):
	def speak(self):
		print("Hello My name is Junior Matin Junior")
```

```python
def attributes_provider(objects):
	return objects.speak()
```

```python
grand = Grandparent()
```

```python
print(attributes_provider(grand))
```

```python
par = Parent()
```

```python
attributes_provider(par)
```

```python
chil = Child()
```

```python
attributes_provider(chil)
```

```python
class Shape:
	def area(self):
		return "this Method return the area of the shape"
```

```python
class Reactangle(Shape):
	def __init__(self,width,height):
		self.width = width
		self.height = height
	def area(self):
		return self.width * self.height
```

```python
class Circle(Shape):
	def __init__(self, radius):
		self.radius = radius

	def area(self):
		return 3.14 * self.radius * self.radius
```

```python
shape = Shape()
shape.area()
```

```python
react = Reactangle(6,4)
react.area()
```

```python
cir = Circle(4)
cir.area()
```
