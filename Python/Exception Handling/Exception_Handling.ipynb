## except ---> which is used to handle the error in way such that program exceution shouldn't stop

## else ---> it will be consisting of the code which we want to run if the try block operation is working without throwing error

## finally ---> Finally block get executed once we are done with everything and even if we are getting an error or not getting an error still the code inside the the finally block gets executed

```python
num = int(input("enter a denominator value : "))
try:
    res = 1 / num 

except Exception as ex:
    print(ex)

else:
    print("the division result for your denominator value will be : " , res)
```
