# Operators

# Arithmetic Operators

```python
# Addition operator
a = 10
b = 20
print(a + b)

# Substraction Operator
print(a - b)

# Multiplication Operator
print(a * b)

#Division Operator
print(a / b)

# Modulus Operator
print(a % b)

# Exponentiation Operator
print(a ** b)
```

# Comparison Operator


```python
a = 10

print(a == 10)

# notEqual operator
print(a != 10)

# Greater than operator
print(a > 10)

# Less than operator
print(a < 10)

# Greater than or equal to operator
print(a >= 10)

# Less than or equal to operator
print(a <= 10)
```
