# Conditional Statement

## conditional statement work like real life scenario where if a certain condition is meeting then a statement of code or a block of code gets executed


```python
# Eligibility for driving on meeting a certain condition
# conditional statement :- if statements where if a condition is meeting then a block of code gets executed

age = int(input("Enter Your Age Value Please : "))

if age  >= 18:
    print("Yes Your eligible for driving")

else:
    print("Sorry! Your are not eligible for driving")
```

# Conditional Statements
# 1. if Statements
# 2. elif Statements
# 3. else Statements

```python
# figure out who is the tallest among the two friends

amit_height = int(input("Enter your Height : "))
raju_height = int(input("Enter your Height : "))

if amit_height > raju_height:
    print("Amit is the tallest among two friends")
else : 
    print("Raju is the tallest among two friends")
```

```python
# figure out who is the tallest among the three friends
amit_height = int(input("Enter the Height Amit : "))
rahul_height = int(input("Enter the Height Rahul : "))
raju_height = int(input("Enter the Height Raju : "))

if amit_height > rahul_height > raju_height:
    print("Amit is the tallest among three friends")
elif rahul_height > amit_height > raju_height :
    print("Rahul is the tallest among the three friends")
else:
    print("Raju is the tallest among the three friends")
```

# Nested Conditional Statement

## Nested Conditional Statements involve if statements inside another if statements

```python
# Compare the height of three friends using nestesd if statements
amit_height = int(input("Enter the Height Amit : "))
rahul_height = int(input("Enter the Height Rahul : "))
raju_height = int(input("Enter the Height Raju : "))

if amit_height > rahul_height:
    if rahul_height >  raju_height or amit_height > raju_height:
        print("Amit is the Tallest guy among the three : ")
elif rahul_height > amit_height:
    if  rahul_height > raju_height or amit_height > raju_height:
        print("Rahul is the Tallest guy among the three : ")
else:
    print("Raju is the Tallest guy among the three")
```

# figure it out weather current year is leap Year or Not

```python
year = int(input("enter a year : "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("The current year is Leap year ")
        else:
            print("The current year is not a Leap Year")
    else:
        print("The current year is Leap year")
else:
    print("The current year is not leap year")
```
