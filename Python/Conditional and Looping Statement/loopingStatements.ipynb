# Loop Control Statements 

loop control statements are being used for terminating a loop in between when a condition is met or if you want to have some control over your iterating process or while itrating over the series of numbers 

# 1. break :- is our first loop control statement which will help you out to terminate the loop when a certain condition is met

```python
## print the number from 1 to 50 but once number=18 then terminate the loop or do not print either 18 or any number forward

for i in range(51):
    if i == 18:
        break 
    else: 
        print(i, end = " , ")
```

# 2. continue :- continue is our second loop control statement which helps us in iterating over a series of number but helps with skipping the values in between once we met a condition

```python
# print the number from 0 to 50 but only print even numbers for odd numbers skip the values

for i in range(51):
    if i % 2 ==0:
        print(i)
    else:
        continue
```

```python
# print the numbers from 0 to 20 but once the 18 is occurring skip it

for i in range(21):
    if i == 18:
        continue
    else:
        print(i)
```

# 3. pass :- pass statement is being used for doing nothing

```python
# print the numbers from 0 to 20 but once 18 is occurring do nothing

for i in range(21):
    if i == 18:
        print("our program will do nothing since the value is 18 and it wont be getting printed we use pass")
        pass
    
    else:
        print(i)
```

# Nested Loops 

Nested loop :- it is going to have the loop inside the loop

f string will we helping us out with the printing the numbers with the variable values 
presesnt inside present inside the colons so we can directly represent the variable value inside the colons message itself where the variable shown inside the {}

```python
# nested loop can helps us out with iterating over series of number as with the proper combination of iterating number

for i in range(6):
    for j in range(4):
        print(f"i value is : {1} and j value is : {j}")
        
```

```python
# print the sum of 1st n natural numbers where natural number start from n ---> 1,2,3,4

n = int(input("enter the value of n : "))
sum = 0
for i in range(n + 1):
    sum += i
print(f" Sum of first n natural number is : {sum}")
```
