## File Handling

# 1. Read the content from my File in the System

## open() --> opening your file
## first parameter ---> File_name
## second parameter ---> Mode
## Mode ---> it decide the way we in which we are working with this file
## Mode --> r --> Reading the content from the File
## Mode --> w --> Write the content from the file
## Mode --> r+ ---> Read the content first and then Writing the content into the file
## Mode --> w+ --> Writing the content first and then reading from it

## Reading your file
## read() --> Reads the Content from the file

```python
f = open('example.txt','r')
content = f.read()
print(content)
f.close()
```

## Write the content into the file 
## write() --> helps us out in writing the content into the file
## write() --> it writes the content into the text file
## Mode --> w --> content will be overwritten 
## Mode --> a --> content will be added from the end 
## append() --> it adds the data from the end of the file

```python
## write() --> helps us out in writing the content into the file
```

```python
f = open("example.txt","w")
f.write('hello we are now writing the content into the file')
f.close()
```

```python
# Appending the content into our file
f = open('example.txt','a')
f.write('\nhello i am adding this content using append ')
f.close()
```

## with --> it will be used for first of all opening the file as the object and there is no requirement for closing the file as well

```python
with open('example.txt','r') as f:
    condata = f.read()
    print(condata)
```

```python
with open('example.txt','a') as f:
    f.write('i am adding the content')
```

```python
# writing the content into our filr line by line ---> writelines()
```

```python
lines = ['firstline\n','secondline\n','thirdline\n']
with open('example.txt', 'w') as f:
    f.writelines(lines)
```

## Mode for Binary files Write is ---> Write the content into the binary file --> wb
## Mode for Binary files Read is --->or we cab read the content in binary files --> rb

```python
data = b'\x00\x01\x02\x03\x04'
with open('example.txt','wb') as bd:
    bd.write(data)
```

```python
with open('example.txt','rb') as bd:
    data_b = bd.read()
    print(data_b)
```

```python
## read the content from the exaple.txt file and writing the content into the another file
```

```python
with open('source.txt' , 'r') as source:
    content = source.read()

with open('destination.txt' , 'w') as destiantion:
    destiantion.write(content)
```

```python
## Reading the total lines and total words and total characters from a file ---> readlines()
def count_lines(file):
	with open(file,'r') as file:
		lines = file.readlines()
		total_lines = len(lines)
		total_words = sum(len(line.split()) for line in lines)
		total_character = sum(len(line) for line in lines)
	return total_lines,total_words,total_character

file_path = 'total_data.txt'
Total_lines,Total_words,Total_character = count_lines(file_path)
print(f"Total Lines are : {Total_lines} , Total Words are : {Total_words} , Total Character are : {Total_character}")


```

## w+ mode --> first of all as W is their so it states that first Writing operation will be done and then Reading operation will be done
## r+ mode --> vice versa
## seek() - helps us with with reading the content after writing from the starting only

```python
with open('total_data.txt' , 'w+') as write_read_file:
    write_read_file.write("\n we are looking into W+ operation ")
    write_read_file.write("\n it means Writing and then Reading ")
    write_read_file.seek(0)

    content = write_read_file.read()
    print(content)
```
