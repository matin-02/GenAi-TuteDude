# Dictionary Methods

important methods for dict 

1. keys()
2. values() 
3. items()
4. get()

```python
dict2 = {"name" : "Alice" , "fruits" : "apple" , "vegatable": "carrot" , "li" : [1,2,3,4,5] , "num" : 36 , "a" : 36 , "Float" : 36.5, "boolean" : True}
```

```python
dict2
```

```python
# keys() key is method which is used to get all the keys from the dictionary.
print(dict2.keys())
```

```python
# values() values is method which is used to get all the values from the dictionary.
print(dict2.values())
```

```python
# items() items is method which is used to get all the key-value pairs together from the dictionary.
print(dict2.items())
```

```python
# get() get is method which is used to get the value of a specific key from the dictionary.
dict2.get("name")
```

```python
print(dict2.get("li"))
```

# Shallow copy : shallow copy is being used to create the copy of the original dictionary in a way such that the changes that we are going to make with original dictionary wont be getting reflected into our copied dictionary and shallow copy is done copy()

```python
dict3 = {"name" : "Alice" , "age" : 25 , "city" : "Mumbai" , "mobile" : 1234567890 , "qualification" : "B.Tech"}
```

```python
dict4 = dict3
```

```python
dict4
```

```python
dict4["age"] = 30
```

```python
dict3
```

```python
dict4
```

```python
dict5 = dict3.copy()
```

```python
print(dict3)
print(dict4)
print(dict5)
```

```python
dict5["age"] = 25
```

```python
print(dict3)
print(dict4)
print(dict5)
```

# Nested Dictionary

nested dictionary is used for keeping dictionary inside the another dictionary

```python
nd = {"student1":{"name" : "Alice" , "age" : 25 , "city" : "Mumbai" , "mobile" : 1234567890 , "qualification" : "B.Tech"},
      "student2":{"name" : "Bob" , "age" : 30 , "city" : "Pune" , "mobile" : 9876543210 , "qualification" : "M.Tech"},
      "student3":{"name" : "Charlie" , "age" : 35 , "city" : "Delhi" , "mobile" : 1234567890 , "qualification" : "Ph.D"}}

```

```python
nd
```

```python
# looping in the dictionary
for outer_key,outer_value in nd.items():
    print(f"outer key is : {outer_key} and outer value is : {outer_value}")
    for key_1,values_1 in outer_value.items():
        print(f"Inner keys is : {key_1} and inner Values is : {values_1}")
```

# Dictionary Compheresion

dictionary compheresion hepls us with using looping statement along with condiation statement inside our dictionary

Syntax : 
{expression for item in iterable if condition }

```python
com = {x ** 2 for x in range(1,11)}
```

```python
com
```

```python
type(com)
```

```python
dic_com = {x:x ** 2 for x in range(11)}
```

```python
dic_com
```

```python
even_com = {x: x ** 2 for x in range(11) if x % 2 == 0}
```

```python
even_com
```
