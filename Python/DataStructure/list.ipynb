# List Method


1. append() :- method is used to add the element at the end of the list
2. insert() :- method is used to add the element at the specific index of the list
3. remove() :- method is used to remove the element from the list
4. pop()    :- method is used to remove the element from the list based on index
5. index()  :- method is used to find the index of the element in the list
6. count()  :- method is used to find the number of occurrences of the element in the list
7. sort()   :- method is used to sort the list in ascending orderfruits.sort()
8. reverse() :- method is used to reverse the list
9. clear()  :- method is used to remove all the elements from the list
10. max() :- method is used to find the maximum value from the list
11. min() : - method is used to find the smallest value from the list
12. sum() :-  method is used to calculate the sum of all number inside the list

```python
#append() method is used to add the element at the end of the list
fruits.append("pineapple")
print(fruits)
```

```python
# insert() method is used to add the element at the specific index of the list
fruits.insert(2,"papaya")
fruits
```

```python
# remove() method is used to remove the element from the list

fruits.remove("pineapple")
fruits
```

```python
# pop() method is used to remove the element from the list based on index
fruits.pop(2)
fruits
```

```python
# index() method is used to find the index of the element in the list
fruits.index("banana")
```

```python
# count() method is used to find the number of occurrences of the element in the list
fruits.count("pineapple")
print(fruits)
```

```python
# sort() method is used to sort the list in ascending orderfruits.sort()
fruits.sort()
print(fruits)
```

```python
# reverse() method is used to reverse the list
fruits.reverse()
print(fruits)
```

```python
# clear() method is used to remove all the elements from the list
fruits.clear()
print(fruits)
```

```python
# sum() method is used to find the sum of all elements inside the list
l1 = [1,2,3,4,5,6,7]
suml = sum(l1)
print(suml)
```

```python
# min()
minl = min(l1)
print(minl)
```

```python
# max()
maxl = max(l1)
print(maxl)
```

# List Comprehension

```python
# list comprehension is a concise way to create lists in Python. 
# It allows you to generate a new list by applying an expression to each item in an existing iterable (like a list, tuple, or string) 
# and optionally filtering the items based on a condition.
```

## Syntax for list comprehension is
# [expression for  item in iterable if condition]
# expression ---> What we want to print
# for item in iterable ---> looping statement
# if condition ---> condition statement

```python
li = [x for x in range(1,51)]
print(li)
```

```python
square = [x ** 2 for x in range(1,16)]
print(square)
```

```python
square_even = [x ** 2 for x in range(1,16) if x % 2 == 0]
print(square_even)
```

## Function in list compheresion as well

```python
fruits = ["apple","banana", "mangeo", "grapes","orange", "kiwi", "watermelon"]

lens_list = [len(word) for word in fruits]
print(lens_list)
```

# Nested List Compheresion
we will be having the combination of multiple lists in our single list itself

```python
pair = [(i,j) for i in fruits for j in lens_list]
print(pair)
```
