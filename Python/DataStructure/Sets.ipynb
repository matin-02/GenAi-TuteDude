# Sets 

sets are data structure which can store the values inside the curly braces format {}.
and are known for storing the values which are non duplicate. duplicate values are not allowed in sets.
set is consist of the unique values


# creating a set in python

```python
a = set({1,2,3,4})
print(a)
```

```python
type(a)
```

```python
b = {1,2,3,4}
```

```python
type(b)
```
