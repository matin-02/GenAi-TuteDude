```python
# Create a function that converts Temperatur from celcius to fahrenheit or vice versa
# fahrenhite to celcius

# if temp in celcius, then celcius to fahrenheit to conversion is :
# 1 deg celcius  = 9/5 + 32
# 45 deg cel = 45 * 9/5 + 32 f

# 1 fah cel = 5/9 - 32
# 45 deg fah = 45 * 5/9 - 32 deg c

def Temperature_change(temp_value , unit):
	if unit =='C' or unit == "c":
		temper = temp_value * 9/5 + 32
		print(f"The temperature for {temp_value} deg celcius in  Fahrenheit is : " , str(temper) + " F")

	elif unit == "F" or unit == "f":
		temper = round((temp_value - 32) * 5/9 , 2) 
		print(f"The temperature for {temp_value} deg Fahrenheit in Celcius is : " , str(temper) + " F")

	else:
		print("Invalid Value Entered By User")
```

```python
Temperature_change(1, 'F')
```

```python
Temperature_change(2,"c")
```

```python
# define the function that waould behave as a password Strength Checker
'''
Strong Password  --->
if Your password:
	1. length is greater than 8 character
	2. password must contain the lower case letter(a,b,c)
	3. password must conatain the upper case letter(A,B,C)
	4. password must consists of atleast a digit(1,2,3,..)
	5. password must consists of atleast a special character(!,@,#,$,)

Weak Passowrd --->
if a password doesn't match any above condition then it will be considered as a weak password

'''
```

'''
islower() :- check that weather if the current character is a lower-case character or not
isupper() :- check that weather if the current character is a upper-case character or not
isdigit() :- check that weather if the current character is a digit or not
'''

```python
def password_Strength(password):
	special_char = '!@#$%^&*()_'
	if len(password) < 8:
		print(f"The entered password {password} is a weak Password ")

	elif not any(char.islower() for char in password):
		print(f"The Entered Password {password} is a weak Password ")

	elif not any(char.isupper() for char in password):
		print(f"The Entered Password {password} is a weak Password ")

	elif not any(char.isdigit() for char in password):
		print(f"The Entered Password {password} is a weak Password ")

	elif not any(char in special_char for char in password):
		print(f"The Entered Password {password} is a week Password ")

	else:
		print(f"The Entered Password {password} is a Strong Password ")
```

```python
password_Strength("Hello")
```

```python
password_Strength("HelloMatin12")
```

```python
password_Strength("helloMatin@12")
```

## palindrome string 
palindrome string is going to have the character similar to its both the side or even yoou will reverse the string you will getting the given string itself

# palindrome String  ---> aba reverse aba
# not palindrome reverse String is not same ---> Ab, abaa, Aba is not same as aba

# revering a String can be done by [::-1]

```python
# Check that Wheather the given String is Palindrome or not

def palindrome(string):
	string_value = string.lower().replace(" ","")

	if string_value == string_value[::-1]:
		print(f"the Current String {string_value} is Palindrome")

	else:
		print(f"The Current String {string_value} is not Palindrome")
```

```python
palindrome("aba")
```

```python
palindrome("abc")
```

```python
palindrome(" A b a")
```

```python
palindrome("A man a plan a canal panama")
```
