# Lambda Function :

### Lmabada function are the one linear functions defined by the Lambda keyword

### Syntax for the Lambda Function is :

### lambda arguments:expression

## lambda ---> keyword which is simplly let the python interpreter know that the current is our lambda function

## Arguments ---> Input values for our function

## Expression  ----> Logic (Function Body)

#### lambda x:x ** 2

### They can have any number of Arugment but only one expression or only one logic

### they are used for short operation or as arguments to higher order functions 



### it can treated as a function without a Name and they dont even require a Return statement 

```python
# function which calculates the square of the numbers 

def square(num):
    return num ** 2
```

```python
print(square(2))
```

```python
# solving using the lambda function
lam_func = lambda x:x**2
```

```python
type(lam_func)
```

```python
lam_func(5)
```

```python
# lambda function to solve the sum of two numbers
sum = lambda x,y: x + y
print(sum(5,6))
```

```python
pro = lambda a,b,c,d: a * b * c * d
print(pro(3,4,2,1))
```
