# Recursive Function

# By a Recursive function we can say that a Function is going to call itself again and again

```python
# factorial of the Number
# factorial of 0 or factorial of 1 must be  ----> 1

def factorial(num):
    if num == 0 or num == 1:
        print(f"the Factorial of {num} is : 1")
    

    else:
        fact = 1
        for i in range(1 , num + 1):
            fact *= i
        print(f"the Factorial of {num} is {fact}")  
```

```python
num = int(input("Enter a Number of the Factorial"))
factorial(num)
```

```python
# factorial recursive
def factorial_recursive(num):
    if num == 0 or num == 1:
        return 1
    else:
        return num * factorial_recursive(num - 1)
```

```python
factorial_recursive(5)
```

```python
number = int(input("Entera a Number"))
factorial_recursive(number)
```

# function with Multiple Parameter :-
in the function with multiple parameters , we are going to pass multiple values as our parameter or more then one parameter we can pass here


```python
# function for addition of numbers
def add(num1 , num2):
    print(f"the addition if two number {num1} + {num2} is {num1 + num2}")
```

# Function with variable length Arguments

## positional arguments  ---> Position arguments are being used so that we can pass multiple values inside our function as an agrument irrespective of the number of values we have mentioned as the parameter inside our function 

```python
# function to add multiple numbers irrespective of how many numbers we are passing

def add_positional(*num):
    add = 0
    for i in num:
        add += i
    print("the addition of numbers are : ", add)
```

```python
add_positional(1,2)
```

```python
add_positional(2,3,4)
```

```python
add_positional(12,34,5,5,66,6,53,2,4,5,6,6,6,76,5,5,443,)
```

# Keywords arguments :- takes the input as the keywords values irrespective of any number of keywords values you are going to pass it wont matter

```python
def data(**user_data):
    for keys,value in user_data.items():
        print(f"{keys} : {value}")
```

```python
data(name = "Matin" , age = 25 , gender = "M")
```

```python
# we want pass the key-value pairs as your input along with positional argument as well
def variable(*positional , **keyword):
    for i in positional:
        print(f"positional values are " , i)

    for keys,value in keyword.items():
        print(f"key is {keys} and the values is {value}") 
```

```python
variable(1,2,3,4,5, name = "amit" , age = 25 ,gender = "male")
```
