```python
# Creating the function to take the input and check weather the number is even or odd 

def even_odd(num):
	if num % 2 == 0:
		 output = "Even Number"
	
	else:
		output = "Odd Number"

	return output
```

```python
def even_odd_instance(num):
	if isinstance(num,int):
		if num % 2 == 0:
			print(f"{num} is Even")
		else:
			print(f"{num} is Odd")
	elif isinstance(num,str):
		print("Invalid data type")
	else:
		print(f"{num} is an invalid value for this function ")
```

```python
even_odd_instance(5)
```

```python
even_odd_instance('s')
```
