# Numpy

### Numpy is the fundamental package for scientific computing in Python.It is Python library that provides a Muti Dimenstion Array oject and various derived obejcts such as Masked Arrays or Matrices and a assortment of routines for fast operation in Arrays including Mathematical , Logical , Shape Manipulation , Sorting , Selecting , I/O , Discrete Fourier Transform , Basic Linear Algebra , Stastical Opreation , Random Simulation , Etc..

### At the core of the Numpy Package is the n-dimensinal Array object.

#### Numpy --> C language ---> Wrapped ---> Python Language

#### Numpy ---> Syntax ---> Easy for beginners --> peforming scientic calculation  ---> creatng multi Dimensional Arrays ---> Speed is faster for applying Operations ---> Less Memory

#### Library like Scikit Learn , Tersorflow and Pytorch require input data in Numpy Arrrays and Tensors

## Numpy Arrays Vs Python Sequences

#### Numpy Arrays have a fixed sized at creation , unlike Python lists which grow dynamically changing the size of an n-dime array will create a new array and delete the original.

#### The elements in Numpy Array are required to be of the same data type 

#### Numpy Arrays facilitate advanced Mathematical and other types of Operations on large number of data . Typically , such operations are executed more effeicienlty with less code than is possible using python's built in sequences
