### Iterating over an Array

```python
array = np.arange(1,17).reshape(4,4)
```

```python
for i in array:
    print(i)
```

```python
new_3d_array = np.arange(1,28).reshape(3,3,3)
```

```python
new_3d_array
```

```python
for i in new_3d_array:
    print(i)
```

### nditer() --> n dimension iteration function it can help us out with iterating over value in a nd array

```python
for i in np.nditer(new_3d_array):
    print(i)
```

### 1D Array --- >  values by value on iteration
### 2D Array ---> Row by Row on iteration
### 3D Array ---> 2d Array by 2D Array
