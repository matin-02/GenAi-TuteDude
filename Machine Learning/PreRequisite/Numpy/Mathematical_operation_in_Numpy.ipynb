### axis ---> it provides us with the idea about the max and min value of anh oeration value inside a specific row or cloumn or we can get a specific value row wise or column wise

#### axis = 0 ---> column wise operation
#### axis = 1 ---> Row wise operation

```python
# Cloumn wise maximum value
np.max(array_data,axis = 0)
```

```python
# Row wise maximum value
np.max(array_data,axis  = 1)
```

```python
np.min(array_data)
```
