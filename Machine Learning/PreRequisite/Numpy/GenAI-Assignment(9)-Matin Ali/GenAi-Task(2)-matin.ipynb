# Task 2 : - Important Mathematical Operations

```python
import numpy as np
```

```python
# Given two arrays:
A = np.array([10,20,30,40])
B = np.array([1,2,3,4])
```

```python
# perform and print the results for : 
# 1 Addititon (A + B)
addi = A + B
print(f"The Addition of Two Arrays {A} and {B} is : \n{addi}")
```

```python
# 2 Substarction ( A - B)
sub = A - B
print(f"The Substraction of Two Arrays {A} and {B} is : \n{sub}")
```

```python
# 3 Multiplication ( A * B)
mul = A * B
print(f"The Multiplication of Two Arrays {A} and {B} is : \n{mul}")
```

```python
# 4 Division (A / B)
div = A / B
print(f"The Division of Two Arrays {A} and {B} is : \n{div}")
```

```python
# Power (A ** 2)
pow = A ** 2
print(f"The Power of {A} is : \n{pow}")
```

```python
# Extra (optional) : Use NumPy function : 
# np.add() , np.substract() from the same operation
print(f"The Addition of Two Arrays {A} and {B} Using NumPy Function is : \n{np.add(A ,B)}")
print(f"The Substraction of Two Arrays {A} and {B} Using NumPy Function is : \n{np.subtract(A ,B)}")
```
