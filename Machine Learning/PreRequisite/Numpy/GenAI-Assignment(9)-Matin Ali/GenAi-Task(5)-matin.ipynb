# Task 5 :- Statistical Operations(Core Focus)

```python
import numpy as np
```

```python
# Given : 
marks = np.array([78 , 85, 90, 66, 72, 88, 95, 60])
```

```python
# 1. Mean
print("Means Value : " , np.mean(marks))
```

```python
# 2. median
print("Meadian Value : " , np.median(marks))
```

```python
# 3. Variance
print("Variance : " , np.var(marks))
```

```python
# Standard Deviation
print("Standard Deviation : " , np.std(marks))
```

```python
# 6. Range(max - min)
print("Range : " , np.max(marks) - np.min(marks))
```
