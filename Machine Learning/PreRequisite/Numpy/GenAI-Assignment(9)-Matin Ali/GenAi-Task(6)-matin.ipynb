# Task 6: Pecentiles and Sorting

```python
import numpy as np
```

```python
marks = np.array([78 , 85, 90, 66, 72, 88, 95, 60])
```

```python
# 1. sort the array
print("Sorted Marks : " , np.sort(marks))
```

```python
# 2. find 
# 25th percentile
print("25th Percentile : " , np.percentile(marks,25))

# 50th percentile
print("50th Percentile : " , np.percentile(marks,50))

# 75th percentile
print("75th Percentile : " , np.percentile(marks,75))
```

```python
# 3 Count how many student scored above the average marks
average_marks = np.mean(marks)
print("Average Marks : " , average_marks) 

above_average_marks = len(marks[marks > average_marks])
print("Number of students scored above average marks : " , above_average_marks)
```
