# Task 7 :- Mini Use Case : Sale Analysis

```python
import numpy as np
```

```python
# Given daily sales data
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
```

```python
# Perform : 
# 1. Total weekly sales
print("Toal Weekly Sales : " , np.sum(sales))
print()

# 2. Average daily sales
print("Average Daily Sales : " , np.mean(sales))
print()

# 3. Highest and Lowest sales day
print("Highest Sales Day : " , np.max(sales) , "\nand The Lowest Sales Day : " , np.min(sales))
print()

# 4. Standard deviation of sales
print("The Standard deviation of Sales : " , np.std(sales))
print()

# 5. Identify days where sales were above average
above_average = len(sales[sales > np.mean(sales)])
print("Sales Above Average : " , above_average)
```
