# Task 4: Aggregation Operations

```python
import numpy as np
```

```python
# Given a 2D array:
data = np.array([[10 , 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])
```

```python
# Find.
# 1 Row-wise sum
print("The Row_wise Sum of 2D Array is : ", np.sum(data,axis = 1))
```

```python
# 2. Column-Wise Sum
print("The Column-wise Sum of 2D Array is : " , np.sum(data,axis = 0))
```

```python
# 3 Minimum value
print("The Minimum value of 2D array is : " , np.min(data))
```

```python
# 4 Maximum value 
print("The Maximum value of 2D Array is : " , np.max(data))
```

```python
# 5.Overall Mean
print("The Overall Mean is : " , np.mean(data))
```
