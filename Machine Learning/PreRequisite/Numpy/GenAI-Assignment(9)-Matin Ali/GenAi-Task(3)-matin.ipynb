# Task 3. Important NumPy Mathematical Formulas

```python
import numpy as np
```

```python
# Given : 
values = np.array([2 ,4 ,6 ,8 ,10])
# calulate : 
# square root if each element
print("The Square root of each element is: ", np.sqrt(values))
print()
# 2. Exponential of each element
print("The Exponential of each element is : ", np.exp(values))
print()
# 3. Natural Logarithm of each elemens=t
print("The Natural Logarithm of each element is : " , np.log(values))
print()
# 4. sum of all elements
print("The Sum of all elements is : ", np.sum(values))
print()
# 5 . Cumulative sum of elements
print("The Cumulative sum of elements is :" , np.cumsum(values))
```
