# Task 1 : Creating NumPy Arrays

```python
# 1. Import NumPy as np
import numpy as np
```

```python
# 2. Create the following arrays:
# 2.1 A 1D arrys of integers from 1 to 10
array_1d = np.arange(1 , 11)
```

```python
# 2.2 A 2D array of shape(3,3) with values from 1 to 9
array_2d = np.arange(1,10).reshape(3,3)
```

```python
# 2.3 A NumPy array from the list: [10,20,30,40,50]

numpy_array = np.array([10,20,30,40,50])
```

```python
# print
# 3.1 shape of each array
print("Shape of 1D Array is : ",array_1d.shape)
print("Shape of 2D Array is : ",array_2d.shape)
print("Shape of Numpy Array from list is : ",numpy_array.shape)
```

```python
# 3.2 data type of each array
print("Data Type of 1D Array is : ",array_1d.dtype)
print("Data Type of 2D Array is : ",array_2d.dtype)
print("Data Type of Numpy Array from list is  : ",numpy_array.dtype)
```
